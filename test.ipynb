```python
from ultralytics import YOLO
import shutil
from pathlib import Path

model = YOLO('models/yolo26m.pt')

results = model.train(
    data='dataset/data.yaml',

    epochs=65,
    imgsz=640,
    batch=24,
    workers=8,
    device='cuda',

    amp=True,
    cache=True,

    lr0=0.002,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,
    cos_lr=True,

    hsv_h=0.015,
    hsv_s=0.6,
    hsv_v=0.4,
    translate=0.08,
    scale=0.4,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.1,
    close_mosaic=10,

    label_smoothing=0.05,

    pretrained=True,
    patience=15,

    name='yolo_dog_cat_l4',
    save=True,
    save_period=10,
)

best_model_path = Path(results.save_dir) / 'weights' / 'best.pt'
final_model_path = Path('model/yolo_dog_cat_best.pt')
final_model_path.parent.mkdir(parents=True, exist_ok=True)

shutil.copy(best_model_path, final_model_path)
```
